### Definicion de funciones

```python
import pandas as pd
from pathlib import Path

def extraer_columnas(
    nombre_archivo: str,
    columnas: list,
    ruta_raw: str = "../data/raw",
    ruta_interim: str = "../data/interim"
) -> Path:
    """
    Lee un CSV desde la carpeta raw, extrae columnas específicas
    y guarda el resultado en la carpeta interim.
    
    """
    ruta_entrada = Path(ruta_raw) / f"{nombre_archivo}.csv"
    ruta_salida = Path(ruta_interim) / f"{nombre_archivo}_columnas.csv"

    df = pd.read_csv(ruta_entrada)

    if columnas is not None:
        columnas_faltantes = set(columnas) - set(df.columns)
        if columnas_faltantes:
            raise ValueError(f"{nombre_archivo}: columnas no encontradas {columnas_faltantes}")
        df = df[columnas]

    df.to_csv(ruta_salida, index=False)


    return ruta_salida
```
