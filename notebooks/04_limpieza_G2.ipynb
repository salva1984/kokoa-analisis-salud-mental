#Limpieza de DR1TOT_H

```python
# from google.colab import drive
# from google.colab import files
# drive.mount('/content/drive')
import pandas as pd
import numpy as np

dr1 = pd.read_csv("../data/interim/dr1tot_h_columnas.csv")
# dr1 = pd.read_csv("/content/drive/MyDrive/kokoa-analisis-salud-mental-codespace-refactored-xylophone-g4566pggg7r9hw9jv/data/interim/dr1tot_h_columnas.csv")
dr1.head()
```

```python
DR1_COLS = dr1.columns.tolist()
dr1 = dr1[DR1_COLS].copy()
dr1.head()
```

```python
codigos_invalidos = [7, 9, 77, 99, 777, 999, 7777, 9999]
dr1.replace(codigos_invalidos, np.nan, inplace=True)

dr1.isna().sum()
```

```python
for col in dr1.columns:
    if col != "SEQN":
        dr1[col] = pd.to_numeric(dr1[col], errors="coerce")

dr1.dtypes
```

```python
dr1.describe()
```

```python
for col in dr1.columns:
    if col != "SEQN":
        dr1[col] = dr1[col].fillna(0)

dr1.isna().sum()
```

```python
conteo_no_cero = ((dr1 != 0)).sum()
```

```python
conteo_no_cero
```

```python
dr1.to_csv('../data/processed/dr1tot_limpio.csv')
print("DR1TOT limpio guardado")
```

#Limpieza de DUQ_H

```python
duq = pd.read_csv('../data/interim/duq_h_columnas.csv')
duq.head()
```

```python
duq.info()
```

```python
DUQ_COLS = duq.columns.tolist()
duq = duq[DUQ_COLS].copy()
duq.head()
```

```python
codigos_invalidos = [7, 9, 77, 99, 777, 999]
duq.replace(codigos_invalidos, np.nan, inplace=True)

duq.isna().sum()
```

```python
for col in duq.columns:
    if col != "SEQN":
        duq[col] = pd.to_numeric(duq[col], errors="coerce")

duq.dtypes
```

```python
for col in duq.columns:
    if col != "SEQN":
        if duq[col].dropna().nunique() == 2:
            duq[col] = duq[col].fillna(duq[col].mode()[0])
        else:
            duq[col] = duq[col].fillna(duq[col].median())

duq.isna().sum()
```

```python
duq.to_csv('../data/processed/duq_limpio.csv')
print("DUQ limpio guardado")
```

#Limpieza de DBQ_H

```python
dbq = pd.read_csv("../data/interim/dbq_h_columnas.csv")
df_demo = pd.read_csv("../data/interim/demo_h_columnas.csv")
df_edad = df_demo[['SEQN', 'RIDAGEYR']].copy()
dbq = dbq.merge(df_edad,on='SEQN',how='inner')
dbq = dbq[dbq['RIDAGEYR']>=18]
dbq.info()
```

```python
# 1. Redondear la columna DBD895 para eliminar los errores de coma flotante
# Redondeamos y la convertimos temporalmente/permanentemente para que el 5.39e-79 sea 0
dbq['DBD895'] = dbq['DBD895'].round()


# si no comiste fuera de casa, no tiene sentido preguntar dbd900
dbq['DBD900'] = np.where(dbq['DBD895'] == 0, 0, dbq['DBD900'])
dbq['DBD900'] = np.where(dbq['DBD895'] == 7777, 7777, dbq['DBD900'])
dbq['DBD900'] = np.where(dbq['DBD895'] == 9999, 9999, dbq['DBD900'])

# 3. Comprobamos los nulos
print(dbq.info())
```

```python
import pandas as pd
import numpy as np


# 2. Arreglar nulos en DBD905 y DBD910
# Tienes muy pocos nulos aquí (14 en DBD905 y 11 en DBD910 de un total de 6113 filas).
# Rellenarlos con la mediana
median_905 = dbq['DBD905'].median()
median_910 = dbq['DBD910'].median()
dbq['DBD905'] = dbq['DBD905'].fillna(median_905)
dbq['DBD910'] = dbq['DBD910'].fillna(median_910)

# 3. Verificación
print(dbq.info())
```

```python
dbq.to_csv('../data/processed/dbq_limpio.csv')
dbq.drop(columns='RIDAGEYR')
print("DBQ limpio guardado")
```
