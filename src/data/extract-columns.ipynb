```python
import pandas as pd

# Ruta del archivo CSV original
archivo_entrada = "ocq_h"

# Leer el CSV
df = pd.read_csv("../../data/raw/" + archivo_entrada + ".csv")

# Ver columnas disponibles (opcional)
print(df.columns)

# Columnas que quieres extraer
columnas_a_extraer = ["SEQN","OCQ180","OCQ210","OCD231","OCD241","OCQ260","OCD390G","OCD391","OCD392","OCD395"]




# Extraer columnas
df_filtrado = df[columnas_a_extraer]

# Guardar en un nuevo CSV
ruta_salida = "../../data/interim/" + archivo_entrada + "_columnas.csv"
df_filtrado.to_csv(ruta_salida, index=False)

print("Archivo guardado correctamente:", ruta_salida)
```
